```python
import os, httpx2

response = httpx2.post(
    "https://api.openai.com/v1/chat/completions",                       # the URL
    headers={"Authorization": f"Bearer {os.environ['OPENAI_API_KEY']}"},  # -H
    json={                                                               # -d (and sets Content-Type for you)
        "model": os.environ["OPENAI_MODEL"],
        "messages": [{"role": "user", "content": "What is the capital of France? One word."}],
    },
    timeout=30,
)
print(response.status_code)                         # like -w '%{http_code}'
data = response.json()                              # like piping to jq
print(data["choices"][0]["message"]["content"], data["model"])
```
